```python
from pathlib import Path

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".webp", ".bmp"
}

print("=" * 80)
print("DATASET IMAGE / LABEL CHECK")
print("=" * 80)

for info in dataset_info:

    dataset = info["yaml"].parent

    print("\n" + "=" * 80)
    print("DATASET:", dataset.name)
    print("=" * 80)

    for split in ["train", "valid", "test"]:

        split_dir = dataset / split

        if not split_dir.exists():
            print(f"{split:5} | ❌ folder not found")
            continue

        image_dir = split_dir / "images"
        label_dir = split_dir / "labels"

        # Find images
        if image_dir.exists():
            images = [
                p for p in image_dir.iterdir()
                if p.is_file()
                and p.suffix.lower() in IMAGE_EXTENSIONS
            ]
        else:
            images = []

        # Find labels
        if label_dir.exists():
            labels = list(label_dir.glob("*.txt"))
        else:
            labels = []

        # Match image and label filenames
        image_stems = {p.stem for p in images}
        label_stems = {p.stem for p in labels}

        missing_labels = image_stems - label_stems
        missing_images = label_stems - image_stems

        print(
            f"{split:5} | "
            f"Images: {len(images):5} | "
            f"Labels: {len(labels):5} | "
            f"Missing labels: {len(missing_labels):4} | "
            f"Missing images: {len(missing_images):4}"
        )
```

```python
print("=" * 80)
print("ANNOTATION FORMAT CHECK")
print("=" * 80)

for info in dataset_info:

    dataset = info["yaml"].parent

    detection_lines = 0
    polygon_lines = 0
    invalid_lines = 0
    empty_files = 0

    label_files = list(dataset.rglob("labels/*.txt"))

    for label_file in label_files:

        try:
            text = label_file.read_text(
                encoding="utf-8"
            ).strip()

            if not text:
                empty_files += 1
                continue

            for line in text.splitlines():

                values = line.strip().split()

                if len(values) == 5:
                    detection_lines += 1

                elif len(values) > 5:
                    polygon_lines += 1

                else:
                    invalid_lines += 1

        except Exception:
            invalid_lines += 1

    print("\n" + "-" * 80)
    print("DATASET:", dataset.name)
    print("Label files:", len(label_files))
    print("Detection lines:", detection_lines)
    print("Polygon lines:", polygon_lines)
    print("Invalid lines:", invalid_lines)
    print("Empty files:", empty_files)
```

```python
from collections import Counter

print("=" * 80)
print("SOURCE CLASS FREQUENCY")
print("=" * 80)

for info in selected_info:

    dataset = info["yaml"].parent
    names = info["names"]

    counter = Counter()

    label_files = list(dataset.rglob("labels/*.txt"))

    for label_file in label_files:

        try:
            text = label_file.read_text(
                encoding="utf-8"
            ).strip()

            if not text:
                continue

            for line in text.splitlines():

                values = line.split()

                if not values:
                    continue

                try:
                    class_id = int(values[0])

                    if 0 <= class_id < len(names):
                        counter[names[class_id]] += 1

                except ValueError:
                    continue

        except Exception:
            continue

    print("\n" + "-" * 80)
    print(dataset.name)

    for class_name, count in counter.most_common():
        print(f"{count:6}  {class_name}")
```

```python
from pathlib import Path
import shutil
import hashlib
import yaml

PROJECT = Path(r"C:\Users\palla\Desktop\Donation_AI")
DATASETS = PROJECT / "datasets"
FINAL = PROJECT / "final_dataset"

USE_DATASETS = [
    "Clothing Detection.v6i.yolov8",
    "First Aid Kid.v1i.yolov8",
    "Furniture Focused Indoor Dataset.v1i.yolov8",
    "Grocery Items.v1-groceryv1.yolov8",
    "Kitchen Utensilsss.v1i.yolov8",
    "Objects -mainly toys- 2.v1i.yolov8",
    "Stationary Items Dataset.v8i.yolov8",
]

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".webp", ".bmp"
}

# Delete any previous incomplete final dataset
if FINAL.exists():
    shutil.rmtree(FINAL)

for split in ["train", "valid", "test"]:
    (FINAL / split / "images").mkdir(parents=True, exist_ok=True)
    (FINAL / split / "labels").mkdir(parents=True, exist_ok=True)

stats = {
    split: {
        "images": 0,
        "labels": 0,
        "skipped": 0,
        "invalid": 0,
    }
    for split in ["train", "valid", "test"]
}

print("=" * 80)
print("BUILDING CLEAN DONATION DATASET")
print("=" * 80)

for dataset_name in USE_DATASETS:

    dataset = DATASETS / dataset_name

    print("\n" + "=" * 80)
    print("DATASET:", dataset_name)
    print("=" * 80)

    yaml_files = (
        list(dataset.glob("*.yaml"))
        + list(dataset.glob("*.yml"))
    )

    if not yaml_files:
        print("❌ No YAML found")
        continue

    with open(yaml_files[0], "r", encoding="utf-8") as f:
        yaml_data = yaml.safe_load(f)

    source_names = yaml_data.get("names", [])

    if isinstance(source_names, dict):
        source_names = [
            source_names[k]
            for k in sorted(source_names, key=lambda x: int(x))
        ]

    print("Source classes:", len(source_names))

    for split in ["train", "valid", "test"]:

        image_dir = dataset / split / "images"
        label_dir = dataset / split / "labels"

        if not image_dir.exists() or not label_dir.exists():
            print(
                f"{split:5} | ⚠️ missing images/labels folder"
            )
            continue

        image_files = [
            p for p in image_dir.iterdir()
            if p.is_file()
            and p.suffix.lower() in IMAGE_EXTENSIONS
        ]

        copied = 0
        skipped = 0

        for image_file in image_files:

            label_file = label_dir / f"{image_file.stem}.txt"

            if not label_file.exists():
                skipped += 1
                stats[split]["skipped"] += 1
                continue

            try:
                lines = label_file.read_text(
                    encoding="utf-8"
                ).splitlines()

                new_labels = []

                for line in lines:

                    values = line.strip().split()

                    # Only YOLO detection format:
                    # class x y width height
                    if len(values) != 5:
                        stats[split]["invalid"] += 1
                        continue

                    try:
                        source_id = int(values[0])

                        x = float(values[1])
                        y = float(values[2])
                        w = float(values[3])
                        h = float(values[4])

                    except ValueError:
                        stats[split]["invalid"] += 1
                        continue

                    # Validate source class ID
                    if not (
                        0 <= source_id < len(source_names)
                    ):
                        stats[split]["invalid"] += 1
                        continue

                    source_name = str(
                        source_names[source_id]
                    ).strip()

                    # Ignore classes that are not mapped
                    if source_name not in CLASS_MAP:
                        continue

                    target_name = CLASS_MAP[source_name]

                    if target_name not in CLASS_TO_ID:
                        stats[split]["invalid"] += 1
                        continue

                    target_id = CLASS_TO_ID[target_name]

                    # Validate YOLO coordinates
                    if not (
                        0 <= x <= 1
                        and 0 <= y <= 1
                        and 0 < w <= 1
                        and 0 < h <= 1
                    ):
                        stats[split]["invalid"] += 1
                        continue

                    new_labels.append(
                        f"{target_id} "
                        f"{x:.6f} "
                        f"{y:.6f} "
                        f"{w:.6f} "
                        f"{h:.6f}"
                    )

                # Nothing useful in this image
                if not new_labels:
                    skipped += 1
                    stats[split]["skipped"] += 1
                    continue

                # Unique filename
                unique_text = (
                    dataset_name
                    + "_"
                    + image_file.name
                )

                unique_id = hashlib.md5(
                    unique_text.encode("utf-8")
                ).hexdigest()[:10]

                new_stem = (
                    f"{dataset_name[:20]}_"
                    f"{image_file.stem}_"
                    f"{unique_id}"
                )

                destination_image = (
                    FINAL
                    / split
                    / "images"
                    / (
                        new_stem
                        + image_file.suffix.lower()
                    )
                )

                destination_label = (
                    FINAL
                    / split
                    / "labels"
                    / f"{new_stem}.txt"
                )

                shutil.copy2(
                    image_file,
                    destination_image
                )

                destination_label.write_text(
                    "\n".join(new_labels) + "\n",
                    encoding="utf-8"
                )

                copied += 1
                stats[split]["images"] += 1
                stats[split]["labels"] += 1

            except Exception as e:
                skipped += 1
                stats[split]["skipped"] += 1

        print(
            f"{split:5} | "
            f"Copied: {copied:5} | "
            f"Skipped: {skipped:5}"
        )

print("\n" + "=" * 80)
print("MERGE COMPLETE")
print("=" * 80)

for split in ["train", "valid", "test"]:
    print(
        f"{split.upper():5} | "
        f"Images: {stats[split]['images']} | "
        f"Labels: {stats[split]['labels']} | "
        f"Skipped: {stats[split]['skipped']} | "
        f"Invalid: {stats[split]['invalid']}"
    )
```

```python
from collections import Counter

print("=" * 80)
print("FINAL CLASS DISTRIBUTION")
print("=" * 80)

class_counts = Counter()

for split in ["train", "valid", "test"]:

    label_dir = FINAL / split / "labels"

    for label_file in label_dir.glob("*.txt"):

        try:
            text = label_file.read_text(
                encoding="utf-8"
            ).strip()

            for line in text.splitlines():

                values = line.split()

                if len(values) != 5:
                    continue

                class_id = int(values[0])

                if 0 <= class_id < len(FINAL_CLASSES):
                    class_counts[class_id] += 1

        except Exception:
            pass

print()

for class_id, class_name in enumerate(FINAL_CLASSES):

    print(
        f"{class_id:3} -> "
        f"{class_name:20} : "
        f"{class_counts[class_id]}"
    )
```

```python
MIN_INSTANCES = 50

KEEP_CLASSES = [
    class_id
    for class_id, count in class_counts.items()
    if count >= MIN_INSTANCES
]

KEEP_CLASSES = sorted(KEEP_CLASSES)

print("=" * 80)
print("CLASS FILTER")
print("=" * 80)

print("Original classes:", len(FINAL_CLASSES))
print("Minimum instances:", MIN_INSTANCES)
print("Classes kept:", len(KEEP_CLASSES))
print("Classes removed:", len(FINAL_CLASSES) - len(KEEP_CLASSES))

print("\nKEPT CLASSES")
print("-" * 80)

for old_id in KEEP_CLASSES:
    print(
        f"{old_id:3} -> "
        f"{FINAL_CLASSES[old_id]:20} : "
        f"{class_counts[old_id]}"
    )

print("\nREMOVED CLASSES")
print("-" * 80)

for old_id in range(len(FINAL_CLASSES)):
    if old_id not in KEEP_CLASSES:
        print(
            f"{old_id:3} -> "
            f"{FINAL_CLASSES[old_id]:20} : "
            f"{class_counts[old_id]}"
        )
```

```python
OLD_TO_NEW = {
    old_id: new_id
    for new_id, old_id in enumerate(KEEP_CLASSES)
}

FINAL_CLASSES_CLEAN = [
    FINAL_CLASSES[old_id]
    for old_id in KEEP_CLASSES
]

print("=" * 80)
print("FINAL CLEAN CLASS LIST")
print("=" * 80)

for new_id, name in enumerate(FINAL_CLASSES_CLEAN):
    old_id = KEEP_CLASSES[new_id]

    print(
        f"{new_id:3} -> {name:20} "
        f"(old ID: {old_id})"
    )

print("\nFINAL CLASS COUNT:", len(FINAL_CLASSES_CLEAN))
```

```python
print("=" * 80)
print("CLASS ID VALIDATION")
print("=" * 80)

invalid = 0
total_labels = 0

for split in ["train", "valid", "test"]:

    label_dir = CLEAN_FINAL / split / "labels"

    for label_file in label_dir.glob("*.txt"):

        for line_number, line in enumerate(
            label_file.read_text(
                encoding="utf-8"
            ).splitlines(),
            start=1
        ):

            values = line.split()

            if len(values) != 5:
                print(
                    "❌ Wrong format:",
                    label_file.name,
                    "line",
                    line_number
                )
                invalid += 1
                continue

            try:
                class_id = int(values[0])

                x, y, w, h = map(
                    float,
                    values[1:]
                )

                total_labels += 1

                if not (
                    0 <= class_id < len(FINAL_CLASSES_CLEAN)
                ):
                    print(
                        "❌ Invalid class ID:",
                        class_id,
                        label_file.name
                    )
                    invalid += 1

                if not (
                    0 <= x <= 1
                    and 0 <= y <= 1
                    and 0 < w <= 1
                    and 0 < h <= 1
                ):
                    print(
                        "❌ Invalid coordinates:",
                        label_file.name
                    )
                    invalid += 1

            except Exception:
                invalid += 1

print()
print("Total valid label lines:", total_labels)
print("Invalid entries:", invalid)

if invalid == 0:
    print("✅ ALL LABELS VALID")
else:
    print("❌ FIX INVALID LABELS BEFORE TRAINING")
```

```python
from collections import Counter

clean_counts = Counter()

for split in ["train", "valid", "test"]:

    label_dir = CLEAN_FINAL / split / "labels"

    for label_file in label_dir.glob("*.txt"):

        for line in label_file.read_text(
            encoding="utf-8"
        ).splitlines():

            values = line.split()

            if len(values) == 5:

                class_id = int(values[0])

                if 0 <= class_id < len(FINAL_CLASSES_CLEAN):
                    clean_counts[class_id] += 1

print("=" * 80)
print("CLEAN CLASS DISTRIBUTION")
print("=" * 80)

for class_id, class_name in enumerate(FINAL_CLASSES_CLEAN):

    print(
        f"{class_id:2} -> "
        f"{class_name:20} : "
        f"{clean_counts[class_id]}"
    )
```

```python
import random
import matplotlib.pyplot as plt
from PIL import Image

images = list(
    (CLEAN_FINAL / "train" / "images").glob("*")
)

samples = random.sample(
    images,
    min(10, len(images))
)

for image_path in samples:

    label_path = (
        CLEAN_FINAL
        / "train"
        / "labels"
        / f"{image_path.stem}.txt"
    )

    image = Image.open(image_path).convert("RGB")

    plt.figure(figsize=(9, 7))
    plt.imshow(image)

    width, height = image.size

    if label_path.exists():

        for line in label_path.read_text(
            encoding="utf-8"
        ).splitlines():

            values = line.split()

            if len(values) != 5:
                continue

            class_id = int(values[0])

            x, y, w, h = map(
                float,
                values[1:]
            )

            x1 = (x - w / 2) * width
            y1 = (y - h / 2) * height

            box_w = w * width
            box_h = h * height

            rect = plt.Rectangle(
                (x1, y1),
                box_w,
                box_h,
                fill=False,
                linewidth=2
            )

            plt.gca().add_patch(rect)

            plt.text(
                x1,
                max(0, y1 - 3),
                FINAL_CLASSES_CLEAN[class_id],
                fontsize=10,
                backgroundcolor="white"
            )

    plt.axis("off")
    plt.show()
```
